```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
model = load_model('model.h5')

with open('le.pkl','rb') as file:
     le = pickle.load(file)

with open('oe.pkl','rb') as file:
    oe =  pickle.load(file)

with open('sc.pkl','rb') as file:
    sc = pickle.load(file)
```

```python
# Example input data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
# geo = oe.transform([[input_data['Geography']]]).toarray()
geo = oe.transform([[input_data['Geography']]])
geo_df = pd.DataFrame(geo, columns = oe.get_feature_names_out(['Geography']))
geo_df
```

```python
input_df = pd.DataFrame([input_data])
input_df
```

```python
input_df['Gender'] = le.transform(input_df['Gender'])
input_df
```

```python
input_df = pd.concat([input_df.drop('Geography', axis = 1), geo_df], axis = 1)
input_df
```

```python
input_sc = sc.transform(input_df)
input_sc
```

```python
prediction = model.predict(input_sc)
prediction
```

```python
pred_prob = prediction[0][0]
pred_prob
```

```python
if pred_prob > 0.5:
    print('The customer is likely to churn.')
else:
    print('The customer is not likely to churn.')
```
