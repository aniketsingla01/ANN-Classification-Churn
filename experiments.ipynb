```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
import pickle
```

```python
data = pd.read_csv("Churn_Modelling.csv")
data.head()
```

```python
data.drop(data.columns[0:3], axis = 1, inplace = True)
data
```

```python
le = LabelEncoder()
data['Gender'] = le.fit_transform(data['Gender'])
data
```

```python
from sklearn.preprocessing import OneHotEncoder
oe = OneHotEncoder(sparse_output=False)
geo = oe.fit_transform(data[['Geography']])
geo
```

```python
### It cant work in ann ####
# data['Geography'] = geo.tolist()
# data[['Geography']].head()

```

```python
oe.get_feature_names_out(['Geography'])
```

```python
geo_df = pd.DataFrame(geo, columns = oe.get_feature_names_out(['Geography']))
geo_df
```

```python
data = pd.concat([data.drop('Geography',axis = 1),geo_df],axis = 1)
data.head()
```

```python
with open('le.pkl','wb') as file:
    pickle.dump(le, file)

with open('oe.pkl','wb') as file:
    pickle.dump(oe, file)
```

```python
X = data.drop('Exited', axis = 1)
y = data['Exited']

X_train, X_test, y_train,y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
with open('sc.pkl','wb') as file:
    pickle.dump(sc, file)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
import datetime
```

```python
(X_train.shape[1],)
```

```python
#ANN Build

model = Sequential([
    Dense(64, activation='relu', input_shape= (X_train.shape[1],)),  #Hodden Layer 1
    Dense(32, activation='relu'), # HL2
    Dense(1, activation='sigmoid')    #OP Layer
])

model.summary()
```

```python
import tensorflow as tf
opt = tf.keras.optimizers.Adam(learning_rate = 0.01)
loss = tf.keras.losses.BinaryCrossentropy()
loss
```

```python
model.compile(optimizer=opt, loss='binary_crossentropy', metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard

log_dir = 'logs/fit/' + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tf_callback = TensorBoard(log_dir=log_dir, histogram_freq=1)
```

```python
early_cb = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights= True)
```

```python
history = model.fit(
    X_train, y_train, validation_data = (X_test,y_test), epochs = 100, callbacks = [tf_callback, early_cb]
)
```

```python
model.save('model.h5')
```
